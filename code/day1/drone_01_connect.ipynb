Now we move from the simulator to the real Crazyflie. The laptop talks to the drone through the Crazyradio dongle, and every drone answers to its own radio address: a channel, a data rate and a five-byte address. In the course the channel and the rate are the factory values, 80 and 2M, and the last byte of the address is the label on the sticker of the drone, a hex digit from 1 to F. The drone with label 2 therefore has the address `E7E7E7E702`, and the full URI, as the radio software wants it, is

```
radio://0/80/2M/E7E7E7E702
```

A wrong address commands another group's drone. So before any control loop we find our drone, make it show itself and read what it reports. This notebook does not arm the motors.

### Before you start

* Switch the drone on (the blue LEDs light up) and put it on the desk, in its protective frame.
* Plug the Crazyradio dongle into the laptop.
* Only one program can talk to a drone at a time: close cfclient before running the cells.


````python
import crazyflie
````

## Which drones are in range?

The next cell pings the addresses of the labels 1 to F, and the factory address, on the channel of the course and lists the drones that answer. It takes about a second. Your drone should appear with the label on its sticker; the drones of the other groups appear too. Each line ends with what to enter in the next cell.

If your drone is missing: is it switched on, is the dongle plugged in? A drone that does not answer on this channel can be searched on all channels with `crazyflie.find_drone(2)`, with your label in place of the 2.


````python
drones = crazyflie.find_drones()
````

## Your drone

Enter the label of your drone as it reads on the sticker, `LABEL = 2`; a letter goes in quotes, `LABEL = 'A'`. The address and the URI follow from it. A drone that still has the factory address `E7E7E7E7E7` has `E7` as its last byte, so `LABEL = 'E7'`.


````python
LABEL = None
URI = crazyflie.drone_uri(LABEL)
````

## Connect and check

The next cell opens the radio link, prints the status of the drone and makes its LEDs blink three times. Watch your drone while the cell runs: if a drone on another desk blinks instead, the label and the address do not match and we need to know.

The status lists the battery voltage (fly only above about 3.7 V, charge over USB when it is lower), whether the flow deck is detected (the deck with the two small windows on the underside, which measures height and motion) and the flags of the supervisor, the part of the firmware that decides whether the drone may be armed.


````python
with crazyflie.connect(URI) as scf:
    crazyflie.status(scf)
    crazyflie.identify(scf)
````

## What the drone measures

The flight notebooks hold the height with the downward range sensor of the flow deck. The next cell prints its reading twice a second for eight seconds. Lift the drone slowly to about 30 cm and put it back: the height follows your hand. The second number is the height estimate of the firmware, which combines the sensor with the accelerometer.


````python
with crazyflie.connect(URI) as scf:
    crazyflie.watch_height(scf, duration=8)
````
